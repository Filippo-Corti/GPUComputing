```python
import torch
```

```python
a = torch.randn(2, 4, 5, 4)
b = torch.randn(1, 1, 4, 3) # Broadcasting of first two dimensions to (2, 4)
c = torch.matmul(a, b)
```

```python
c.shape
```

```python
A = torch.tensor([[1.], [2.], [3.], [4.]]) # Column vector
B = torch.tensor([[10., 20., 30.]])   # Row vector

A.shape, B.shape
```

```python
C = A + B # Broadcasting of both dimensions to (4, 3)
C.shape, C
```

### EinSum Notation

```python
A = torch.randn(2,3)
B = torch.randn(2,3)
#C = torch.einsum('ik, kj -> ij', A, B) # Matrix multiplication
D = torch.einsum('ij, ij -> ij', A, B) # Pointwise multiplication (Hadamard product)
```

```python
D.shape, D
```

```python
P = torch.randn(2, 3, 4)
Q = torch.randn(2, 4, 5)
torch.einsum("nij, njk -> nik", P, Q) # Multiplication over the last dimension of P and the second-to-last dimension of Q
```
